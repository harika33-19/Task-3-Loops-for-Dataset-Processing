```python
students = {
    "Ravi": 85,
    "Sita": 90,
    "Ram": 78,
    "Priya": 92,
    "Arjun": 67,
    "Kavya": 88,
    "Manoj": 76,
    "Divya": 95,
    "Kiran": 81,
    "Anu": 73
}
```

```python
for name, marks in students.items():
    print(name, ":", marks)
```

```python
total = 0

for marks in students.values():
    total += marks

average = total / len(students)

print("Total Marks:", total)
print("Average Marks:", round(average, 2))
```

```python
marks_list = list(students.values())

maximum = marks_list[0]
minimum = marks_list[0]

for marks in marks_list:
    if marks > maximum:
        maximum = marks
    if marks < minimum:
        minimum = marks

print("Highest Marks:", maximum)
print("Lowest Marks:", minimum)
```

```python
for name, marks in students.items():
    if marks >= 90:
        result = "Excellent"
    elif marks >= 75:
        result = "Good"
    elif marks >= 50:
        result = "Average"
    else:
        result = "Needs Improvement"

    print(name, "-", marks, "-", result)
```

```python
names = list(students.keys())
i = 0

while i < len(names):
    name = names[i]
    print(name, ":", students[name])
    i += 1
```

```python
marks_list = list(students.values())

total = 0
count = 0
passed = 0
failed = 0

for marks in marks_list:
    total += marks
    count += 1

    if marks >= 50:
        passed += 1
    else:
        failed += 1

average = total / count

print("Total Students:", count)
print("Total Marks:", total)
print("Average Marks:", round(average, 2))
print("Highest Marks:", max(marks_list))
print("Lowest Marks:", min(marks_list))
print("Passed:", passed)
print("Failed:", failed)
```
